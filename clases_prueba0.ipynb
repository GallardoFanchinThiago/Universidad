```python
class Observaciones:

    def _init_(self):
        pass

    datos = {
        "temperatura" : [],
        "presion" : [],
        "humedad" : [],
        "intensidad_viento" : [],
        "direccion_viento" : [],
        "octavos_nubosidad" : [],
        "tipos_nubosidad" : [],
        "precipitacion" : [],
        "tipo_precipitacion" : [],
        }

    unidades = {
        "temperatura" : "°C",
        "presion" : "hPa",
        "humedad" : "%",
        "intensidad_viento" : "kt",
        "direccion_viento" : "°",
        "octavos_nubosidad" : "/8",
        "tipos_nubosidad" : "",
        "precipitacion" : "mm",
        "tipo_precipitacion" : "",
    }

    def agr_temperatura(self,t):
        (self.datos["temperatura"]).append(t)

    def agr_presion(self,p):
        self.datos["presion"].append(p)

    def agr_humedad(self,h):
        self.datos["humedad"].append(h)

    def agr_viento(self,v,dir):
        self.datos["intensidad_viento"].append(v)
        self.datos["direccion_viento"].append(dir)

    def agr_nubosidad(self,n,tipo):
        self.datos["octavos_nubosidad"].append(n)
        self.datos["tipos_nubosidad"].append(tipo)

    def agr_precipitacion(self,pp,tipo):

        tipo_de = {
            0:"Sin precipitación",
            1:"Lluvia",
            2:"Llovizna",
            3:"Chaparrón",
            4:"Nieve",
        }

        self.datos["precipitacion"].append(pp)
        self.datos["tipo_precipitacion"].append(tipo_de[tipo])

    def agregar_observacion(self,data):
        for i,n in zip(self.datos,data):
            self.datos[i].append(n)


    def quitar_observacion(self,id):
        for i in self.datos:
            if self.datos[i]:
                self.datos[i].pop(id)

    def mostrar_observaciones(self):
        for i in self.datos:
            if self.datos[i]:
                print(f"{i.capitalize()} : {self.datos[i]} ({self.unidades[i]})")
```

```python
chepes = Observaciones()
```

```python
import random

def random_data():
    dato = []
    parametros_normales = [(28,4), (1013,10), (40,20)]
    parametros_enteros = [((0,40), (0,360)), ((0,8), (0,9)), ((0,300), (0,4))]
    for i in parametros_normales:
        dato.append(round(random.gauss(*i),2))
    for i in parametros_enteros:
        a,b = (round(random.randint(*a)) for a in i)
        dato.append(a)
        dato.append(b)

    return dato
```

```python
for _ in range(8):
    chepes.agregar_observacion(random_data())
```

```python
chepes.mostrar_observaciones()
```

```python
while any(chepes.datos.values()):
    chepes.quitar_observacion(0)
```
