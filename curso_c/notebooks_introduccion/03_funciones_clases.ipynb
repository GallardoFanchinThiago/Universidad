## 1. Modularización con Funciones
Una función se define con la palabra clave `def`, opcionalmente acepta parámetros de entrada y devuelve un resultado utilizando `return`.

```python
def calcular_punto_rocio(temp, humedad):
    # Una aproximación simple para calcular el punto de rocío
    return temp - ((100 - humedad) / 5)

rocio = calcular_punto_rocio(25.0, 60)
print(f'Punto de Rocío estimado: {rocio}°C')
```

## 2. Ejemplo: Funciones
### Índice de Sensación Térmica (Wind Chill)
La fórmula estandarizada de sensación térmica por efecto del viento para temperaturas menores a 10°C es:
$$T_{wc} = 13.12 + 0.6215 \cdot T - 11.37 \cdot V^{0.16} + 0.3965 \cdot T \cdot V^{0.16}$$
Donde $T$ es la temperatura en °C y $V$ es la velocidad del viento en km/h.

Completa la función `calcular_sensacion_termica` para que reciba $T$ y $V$, y devuelva el valor de sensación térmica redondeado a 1 decimal utilizando `round(resultado, 1)`.

```python
def calcular_sensacion_termica(t, v):
    twc = 13.12 + 0.6215*t -11.37*v**0.16 +0.3965*t*v**0.16 # TU CÓDIGO AQUÍ
    return round(twc, 1)

# Prueba de ejecución
print(f'Sensación térmica calculada (5°C, 20 km/h): {calcular_sensacion_termica(5, 20)}')
```

```python
# CELDA DE VERIFICACIÓN
try:
    ans = calcular_sensacion_termica(5, 20)
    assert ans == 1.1, f'La respuesta para T=5, V=20 debería ser 1.1, obtuviste {ans}'
    assert calcular_sensacion_termica(0, 15) - (-4.4) >= -0.015, 'Error para los parámetros T=0, V=15.'
    print('¡Felicidades! Desafío 3.1 (Función de Sensación Térmica) superado. 💨')
except AssertionError as e:
    print(f'Error de validación: {e}')
```

## 3. Clases y Objetos (Programación Orientada a Objetos)
Las clases nos permiten estructurar datos y comportamiento en una sola entidad. En meteorología, podemos modelar una `EstacionMeteorologica` que guarde su historial de temperaturas.

### Ejercicio: Estructurar la clase EstacionMeteorologica
Completa el método `agregar_temperatura` y el método `promedio_temperaturas` para la clase dada.

```python
class EstacionMeteorologica:
    def __init__(self, nombre):
        self.nombre = nombre
        self.temperaturas = []

    def agregar_temperatura(self, temp):
        self.temperaturas.append(temp)
        pass

    def promedio_temperaturas(self):
        if len(self.temperaturas) > 0:
            return sum(self.temperaturas)/len(self.temperaturas)
        else:
            return 0.0
```

```python
# CELDA DE VERIFICACIÓN
try:
    estacion = EstacionMeteorologica('Córdoba')
    assert estacion.nombre == 'Córdoba', 'El constructor no asignó correctamente el nombre.'
    estacion.agregar_temperatura(18.5)
    estacion.agregar_temperatura(21.5)
    assert len(estacion.temperaturas) == 2, 'El método agregar_temperatura no funcionó correctamente.'
    assert estacion.promedio_temperaturas() == 20.0, 'El promedio calculado es incorrecto.'
    print('Resultado verificado. 🗺️')
except AssertionError as e:
    print(f'Error: {e}')
```

```python
# KOAN: El Camino de la Mutabilidad y las Listas
# Completa las variables '__' para que el Koan sea superado y alcances la iluminación.

__ = None  # Reemplaza esto con el valor correcto en cada aserción

class KoanMetereologia:
    def __init__(self):
        self.sensores = ["Termómetro", "Anemómetro"]

    def agregar_sensor(self, sensor):
        self.sensores.append(sensor)
        return self.sensores

try:
    # Koan 1: El estado inicial de la mente (y del objeto)
    koan = KoanMetereologia()
    # ¿Cuántos sensores iniciales tiene nuestro koan?
    assert len(koan.sensores) == __  # Cambia por el valor correcto si fuera necesario

    # Koan 2: La mutación del camino
    koan.agregar_sensor("Barómetro")
    # ¿Cuál es la longitud actual de la lista de sensores?
    assert len(koan.sensores) == __

    # Koan 3: La referencia compartida
    # Si asignamos la lista de sensores a una nueva variable...
    lista_referenciada = koan.sensores
    lista_referenciada.append("Pluviómetro")

    # ¿Se habrá modificado la lista original dentro de nuestro objeto 'koan'?
    # Reemplaza el '__' con True o False según corresponda:
    koan_iluminado = __  # ¿Es verdadero que koan.sensores ahora tiene "Pluviómetro"?
    assert ("Pluviómetro" in koan.sensores) == koan_iluminado

    print("Has alcanzado la iluminación en el Koan de la Mutabilidad. 🌸")
except AssertionError as e:
    print("Tu mente aún divaga. Sigue meditando en el flujo del código y corrige los valores.")
```

### El Koan del Diccionario Mutable

En Python, los diccionarios también son objetos mutables. Completa el siguiente Koan para entender cómo interactúan las referencias con las estructuras clave-valor.

```python
# KOAN: El Diccionario y sus Claves
# Completa las variables '__' para superar el Koan y comprender la mutabilidad de los mapas.

__ = None  # Marcador de posición

try:
    # Koan 1: El mapa inicial de la estación
    metadatos = {"estacion": "Central", "altitud": 1500}

    # ¿Cuántas claves contiene inicialmente nuestro diccionario?
    assert len(metadatos) == 2

    # Koan 2: La bifurcación del camino (referencias compartidas)
    copia_metadatos = metadatos
    copia_metadatos["responsable"] = "Dra. Ana"

    # Dado que los diccionarios son mutables y compartimos la referencia...
    # ¿Tiene el diccionario original la clave "responsable"?
    # Reemplaza el '__' con True o False:
    existe_responsable = __
    assert ("responsable" in metadatos) == existe_responsable

    # Koan 3: La verdadera copia (Independencia)
    # Si usamos el método .copy(), creamos una copia superficial independiente
    copia_real = metadatos.copy()
    copia_real["altitud"] = 2000

    # ¿Habrá cambiado la altitud en el diccionario original 'metadatos'?
    # Reemplaza el '__' con el valor numérico esperado:
    altitud_original = __
    assert metadatos["altitud"] == altitud_original

    print("Has completado el Koan de los Diccionarios y la Mutabilidad. 🗺️🌸")
except AssertionError as e:
    print("Sigue observando el flujo del mapa de datos, la respuesta correcta está en la naturaleza de las referencias.")
```
